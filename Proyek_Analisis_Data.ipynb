## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

### Assessing Data

#### Identifying Data Quality Problems

```python
print("=" * 80)
print("ASSESSING DATA QUALITY")
print("=" * 80)

# 1. Cek Missing Values
print("\n1️⃣ MISSING VALUES")
print("-" * 80)
missing_day = df_day.isnull().sum()
missing_hour = df_hour.isnull().sum()

print("\n📋 Daily Data - Missing Values:")
print(missing_day[missing_day > 0] if missing_day.sum() > 0 else "✓ Tidak ada missing values")

print("\n📋 Hourly Data - Missing Values:")
print(missing_hour[missing_hour > 0] if missing_hour.sum() > 0 else "✓ Tidak ada missing values")

# 2. Cek Duplicate Data
print("\n2️⃣ DUPLICATE DATA")
print("-" * 80)
print(f"Daily Data - Duplicate rows: {df_day.duplicated().sum()}")
print(f"Hourly Data - Duplicate rows: {df_hour.duplicated().sum()}")

# 3. Cek Data Range dan Distribusi
print("\n3️⃣ DATA RANGE & DISTRIBUSI")
print("-" * 80)

# Cek untuk normalized values (seharusnya 0-1)
normalized_cols = ['temp', 'atemp', 'hum', 'windspeed']
print("\nNormalized Features (harus dalam range 0-1):")
for col in normalized_cols:
    min_val = df_day[col].min()
    max_val = df_day[col].max()
    print(f"  {col}: min={min_val:.4f}, max={max_val:.4f}")
    if min_val < 0 or max_val > 1:
        print(f"    ⚠️  WARNING: Ada nilai di luar range [0, 1]!")

# Cek nilai kategorikal
print("\nCategorical Features Range (Daily Data):")
print(f"  season: {df_day['season'].min()}-{df_day['season'].max()} (expected: 1-4)")
print(f"  weathersit: {df_day['weathersit'].min()}-{df_day['weathersit'].max()} (expected: 1-4)")
print(f"  weekday: {df_day['weekday'].min()}-{df_day['weekday'].max()} (expected: 0-6)")

# 4. Cek Outliers menggunakan IQR method
print("\n4️⃣ OUTLIERS DETECTION (IQR Method)")
print("-" * 80)

def detect_outliers(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = data[(data[column] < lower_bound) | (data[column] > upper_bound)]
    return len(outliers), lower_bound, upper_bound

outlier_cols = ['casual', 'registered', 'cnt']
print("\nDaily Data - Outliers:")
for col in outlier_cols:
    n_outliers, lower, upper = detect_outliers(df_day, col)
    print(f"  {col}: {n_outliers} outliers (bounds: {lower:.2f} - {upper:.2f})")

# 5. Cek Inaccurate Values (cnt = casual + registered)
print("\n5️⃣ INACCURATE VALUES - Consistency Check")
print("-" * 80)
df_day['total_check'] = df_day['casual'] + df_day['registered']
inconsistent = (df_day['cnt'] != df_day['total_check']).sum()
print(f"Daily Data: cnt = casual + registered?")
print(f"  Inconsistent rows: {inconsistent}")
if inconsistent > 0:
    print(f"  ⚠️  WARNING: Ditemukan {inconsistent} baris dengan inkonsistensi!")
    
df_hour['total_check'] = df_hour['casual'] + df_hour['registered']
inconsistent_hour = (df_hour['cnt'] != df_hour['total_check']).sum()
print(f"\nHourly Data: cnt = casual + registered?")
print(f"  Inconsistent rows: {inconsistent_hour}")
if inconsistent_hour > 0:
    print(f"  ⚠️  WARNING: Ditemukan {inconsistent_hour} baris dengan inkonsistensi!")

print("\n" + "=" * 80)
print("DESCRIPTIVE STATISTICS - DAILY DATA")
print("=" * 80)
print(df_day.describe())
```

**Data Quality Issues Ditemukan:**

1. **Outliers dalam Pengguna (Casual & Registered)**: Ditemukan beberapa hari dengan jumlah pengguna yang jauh berbeda dari rata-rata, kemungkinan karena hari libur atau kondisi cuaca ekstrem. Ini adalah data yang valid dan tidak perlu dihapus, melainkan perlu dianalisis lebih lanjut.

2. **Inaccurate Values**: Perlu verifikasi bahwa nilai 'cnt' (total pengguna) = 'casual' + 'registered'. Jika ada inkonsistensi, perlu dilakukan perbaikan.

**Steps to Take:**

### Cleaning Data

#### Data Type Conversion & Consistency Validation

```python
# Convert dteday to datetime
df_day['dteday'] = pd.to_datetime(df_day['dteday'])
df_hour['dteday'] = pd.to_datetime(df_hour['dteday'])

# Validasi consistency: cnt = casual + registered
print("=" * 80)
print("DATA CLEANING")
print("=" * 80)

print("\n✓ Converting dteday to datetime format")
print(f"  Daily Data date range: {df_day['dteday'].min()} to {df_day['dteday'].max()}")
print(f"  Hourly Data date range: {df_hour['dteday'].min()} to {df_hour['dteday'].max()}")

# Verifikasi consistency
invalid_day = df_day[df_day['cnt'] != (df_day['casual'] + df_day['registered'])]
invalid_hour = df_hour[df_hour['cnt'] != (df_hour['casual'] + df_hour['registered'])]

if len(invalid_day) == 0:
    print("\n✓ Daily Data: cnt = casual + registered (VALID)")
else:
    print(f"\n⚠️  Daily Data: Found {len(invalid_day)} inconsistent rows")
    # Fix if needed
    df_day.loc[invalid_day.index, 'cnt'] = df_day.loc[invalid_day.index, 'casual'] + df_day.loc[invalid_day.index, 'registered']
    print(f"  Fixed: Updated cnt values")

if len(invalid_hour) == 0:
    print("✓ Hourly Data: cnt = casual + registered (VALID)")
else:
    print(f"⚠️  Hourly Data: Found {len(invalid_hour)} inconsistent rows")
    # Fix if needed
    df_hour.loc[invalid_hour.index, 'cnt'] = df_hour.loc[invalid_hour.index, 'casual'] + df_hour.loc[invalid_hour.index, 'registered']
    print(f"  Fixed: Updated cnt values")

# Create mapping untuk season dan weather
season_map = {1: 'Spring', 2: 'Summer', 3: 'Fall', 4: 'Winter'}
weather_map = {1: 'Clear', 2: 'Mist', 3: 'Light Snow/Rain', 4: 'Heavy Rain'}
weekday_map = {0: 'Sunday', 1: 'Monday', 2: 'Tuesday', 3: 'Wednesday', 4: 'Thursday', 5: 'Friday', 6: 'Saturday'}

df_day['season_name'] = df_day['season'].map(season_map)
df_day['weather_name'] = df_day['weathersit'].map(weather_map)
df_day['weekday_name'] = df_day['weekday'].map(weekday_map)

df_hour['season_name'] = df_hour['season'].map(season_map)
df_hour['weather_name'] = df_hour['weathersit'].map(weather_map)
df_hour['weekday_name'] = df_hour['weekday'].map(weekday_map)

print("\n✓ Created meaningful season, weather, dan weekday labels")
print(f"  Season labels: {list(season_map.values())}")
print(f"  Weather labels: {list(weather_map.values())}")

# Drop temporary check column
df_day = df_day.drop('total_check', axis=1)
df_hour = df_hour.drop('total_check', axis=1)

print("\n" + "=" * 80)
print("DATA CLEANING - COMPLETED")
print("=" * 80)
print(f"\n✓ Daily Data shape: {df_day.shape}")
print(f"✓ Hourly Data shape: {df_hour.shape}")
print(f"\n✓ Data siap untuk Exploratory Data Analysis")
```

## Exploratory Data Analysis (EDA)

Pada bagian ini, kami akan melakukan eksplorasi mendalam terhadap data untuk menjawab kedua pertanyaan bisnis yang telah dirumuskan.

### Pertanyaan 1: Pola Penggunaan Berdasarkan Musim, Hari Kerja, dan Cuaca

#### Analisis 1.1: Pola Penggunaan Berdasarkan Musim

```python
print("=" * 80)
print("EDA 1.1: POLA PENGGUNAAN BERDASARKAN MUSIM")
print("=" * 80)

# Agregasi berdasarkan musim
season_analysis = df_day.groupby('season_name').agg({
    'casual': ['mean', 'median', 'std', 'min', 'max'],
    'registered': ['mean', 'median', 'std', 'min', 'max'],
    'cnt': ['mean', 'median', 'std', 'min', 'max']
}).round(2)

print("\n📊 Statistik Pengguna per Musim:")
print(season_analysis)

# Hitung persentase pengguna casual vs registered per musim
season_composition = df_day.groupby('season_name')[['casual', 'registered', 'cnt']].mean()
season_composition['casual_pct'] = (season_composition['casual'] / season_composition['cnt'] * 100).round(2)
season_composition['registered_pct'] = (season_composition['registered'] / season_composition['cnt'] * 100).round(2)

print("\n📈 Komposisi Pengguna (Casual vs Registered) per Musim:")
print(season_composition[['casual', 'registered', 'casual_pct', 'registered_pct']])

# Analisis korelasi dengan faktor cuaca per musim
print("\n🌤️  Rata-rata Kondisi Cuaca per Musim:")
weather_per_season = df_day.groupby('season_name')[['temp', 'atemp', 'hum', 'windspeed']].mean().round(3)
print(weather_per_season)

# Anova test untuk melihat signifikansi perbedaan antar musim
from scipy.stats import f_oneway
spring_users = df_day[df_day['season_name'] == 'Spring']['cnt'].values
summer_users = df_day[df_day['season_name'] == 'Summer']['cnt'].values
fall_users = df_day[df_day['season_name'] == 'Fall']['cnt'].values
winter_users = df_day[df_day['season_name'] == 'Winter']['cnt'].values

f_stat, p_value = f_oneway(spring_users, summer_users, fall_users, winter_users)
print(f"\n📊 ANOVA Test (Perbedaan signifikan antar musim):")
print(f"  F-statistic: {f_stat:.2f}")
print(f"  P-value: {p_value:.2e}")
print(f"  Kesimpulan: {'Perbedaan SIGNIFIKAN' if p_value < 0.05 else 'Tidak signifikan'} antar musim")
```

**Insight 1.1: Pola Penggunaan Berdasarkan Musim**
- Summer memiliki rata-rata pengguna tertinggi, diikuti Fall, Spring, dan Winter
- Registered users menunjukkan pola musiman yang lebih jelas dibandingkan casual users
- Musim memiliki pengaruh statistik yang signifikan terhadap jumlah pengguna (p-value < 0.05)

#### Analisis 1.2: Pola Penggunaan Berdasarkan Hari Kerja vs Hari Libur/Weekend

#### Analisis 1.3: Pola Penggunaan Berdasarkan Kondisi Cuaca

```python
print("=" * 80)
print("EDA 1.2: POLA PENGGUNAAN BERDASARKAN HARI KERJA vs WEEKEND/LIBUR")
print("=" * 80)

# Analisis berdasarkan workingday
workingday_analysis = df_day.groupby('workingday').agg({
    'casual': ['mean', 'median', 'std'],
    'registered': ['mean', 'median', 'std'],
    'cnt': ['mean', 'median', 'std']
}).round(2)

workingday_labels = {0: 'Weekend/Libur', 1: 'Hari Kerja'}
print("\n📊 Statistik Pengguna: Hari Kerja vs Weekend/Libur:")
print(workingday_analysis)

# Perbandingan sederhana
workingday_comp = df_day.groupby('workingday')[['casual', 'registered', 'cnt']].mean()
workingday_comp.index = workingday_comp.index.map(workingday_labels)
print("\n📈 Rata-rata Pengguna:")
print(workingday_comp.round(2))

# T-test untuk membandingkan hari kerja vs weekend
from scipy.stats import ttest_ind
working_users = df_day[df_day['workingday'] == 1]['cnt'].values
weekend_users = df_day[df_day['workingday'] == 0]['cnt'].values
t_stat, p_value = ttest_ind(working_users, weekend_users)
print(f"\n📊 T-Test (Perbedaan hari kerja vs weekend):")
print(f"  T-statistic: {t_stat:.2f}")
print(f"  P-value: {p_value:.2e}")
print(f"  Kesimpulan: {'Perbedaan SIGNIFIKAN' if p_value < 0.05 else 'Tidak signifikan'}")

# Analisis per hari dalam seminggu
print("\n" + "=" * 80)
print("ANALISIS PER HARI DALAM SEMINGGU")
print("=" * 80)
weekday_analysis = df_day.groupby('weekday_name')[['casual', 'registered', 'cnt']].agg(['mean', 'median', 'std']).round(2)
print(weekday_analysis)

# Urutkan hari dari Monday sampai Sunday
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
weekday_avg = df_day.groupby('weekday_name')[['casual', 'registered', 'cnt']].mean().reindex(day_order)
print("\n📈 Rata-rata Pengguna per Hari (Terurut):")
print(weekday_avg.round(2))
```

**Insight 1.2: Pola Penggunaan Hari Kerja vs Weekend**
- Hari kerja memiliki rata-rata pengguna yang LEBIH TINGGI dibandingkan weekend/libur
- Registered users menunjukkan perbedaan lebih signifikan antara hari kerja dan weekend
- Casual users lebih konsisten antara hari kerja dan weekend

```python
print("=" * 80)
print("EDA 1.3: POLA PENGGUNAAN BERDASARKAN KONDISI CUACA")
print("=" * 80)

# Analisis berdasarkan weathersit
weather_analysis = df_day.groupby('weather_name').agg({
    'casual': ['mean', 'median', 'std', 'count'],
    'registered': ['mean', 'median', 'std'],
    'cnt': ['mean', 'median', 'std']
}).round(2)

print("\n📊 Statistik Pengguna per Kondisi Cuaca:")
print(weather_analysis)

# Perbandingan rata-rata
weather_comp = df_day.groupby('weather_name')[['casual', 'registered', 'cnt']].mean().sort_values('cnt', ascending=False)
print("\n📈 Rata-rata Pengguna (Urut dari tertinggi):")
print(weather_comp.round(2))

# Anova test untuk cuaca
clear_users = df_day[df_day['weather_name'] == 'Clear']['cnt'].values
mist_users = df_day[df_day['weather_name'] == 'Mist']['cnt'].values
rain_users = df_day[df_day['weather_name'] == 'Light Snow/Rain']['cnt'].values

f_stat, p_value = f_oneway(clear_users, mist_users, rain_users)
print(f"\n📊 ANOVA Test (Perbedaan signifikan antar kondisi cuaca):")
print(f"  F-statistic: {f_stat:.2f}")
print(f"  P-value: {p_value:.2e}")
print(f"  Kesimpulan: {'Perbedaan SIGNIFIKAN' if p_value < 0.05 else 'Tidak signifikan'}")

# Korelasi Pearson antara faktor cuaca dan total users
print("\n" + "=" * 80)
print("KORELASI FAKTOR CUACA DENGAN TOTAL USERS")
print("=" * 80)
weather_factors = ['temp', 'atemp', 'hum', 'windspeed', 'cnt']
correlation = df_day[weather_factors].corr()['cnt'].drop('cnt').sort_values(ascending=False)
print("\nKorelasi Pearson dengan Total Users (cnt):")
for factor, corr_val in correlation.items():
    print(f"  {factor:12s}: {corr_val:7.4f}")
    
# Interpretasi
print("\nInterpretasi:")
print("  • Korelasi positif: Semakin tinggi nilai, semakin banyak pengguna")
print("  • Korelasi negatif: Semakin tinggi nilai, semakin sedikit pengguna")
print("  • Nilai absolut |r| > 0.7: Korelasi KUAT")
print("  • Nilai absolut |r| 0.4-0.7: Korelasi SEDANG")
print("  • Nilai absolut |r| < 0.4: Korelasi LEMAH")
```

**Insight 1.3: Pola Penggunaan Berdasarkan Cuaca**
- Cuaca cerah (Clear) memiliki rata-rata pengguna tertinggi, diikuti Mist dan Light Snow/Rain
- Perbedaan pengguna antar kondisi cuaca secara statistik signifikan
- Faktor cuaca seperti suhu memiliki korelasi positif KUAT dengan total pengguna
- Kelembaban memiliki korelasi negatif dengan pengguna (semakin lembab, semakin sedikit pengguna)

### Pertanyaan 2: Karakteristik Hari Peak vs Low Demand dan Pola Per Jam

  - Weekend: discount 10% untuk meningkatkan casual users
